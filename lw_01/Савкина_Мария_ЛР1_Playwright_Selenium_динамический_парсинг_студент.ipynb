## Теоретический блок 1. Почему статический парсинг бессилен

### Два способа доставить страницу пользователю

| | Server-Side Rendering | Client-Side Rendering (SPA) |
|---|---|---|
| Что приходит в ответ на GET | готовый HTML с данными | «скелет»: пустой `<div id="root">` и ссылки на бандлы |
| Где берутся данные | сервер собрал их до отправки | браузер выполняет JS, тот запрашивает JSON и строит DOM |
| Что увидит `requests` | данные | пустой каркас |
| Типичный стек | Django, Rails, классический PHP | React, Angular, Vue, Svelte |

**Гидратация** — момент, когда загруженный JavaScript «оживляет» разметку и подставляет данные. До неё в DOM нужных узлов физически нет. В нашем стенде это видно буквально: в исходном HTML ноль строк таблицы, а после гидратации их двенадцать.

### Как отличить одно от другого за 15 секунд

1. `view-source:` или `requests.get(url).text` — есть ли в ответе искомое число?
2. Если нет — открыть DevTools → вкладка **Network** → фильтр **Fetch/XHR** и перезагрузить страницу. Часто данные приходят одним JSON-запросом: тогда браузер не нужен вовсе, достаточно обратиться к этому эндпоинту (это самый быстрый и устойчивый путь).
3. Если данные собираются множеством запросов, приходят в зашифрованном виде или защищены проверкой браузера — нужен headless-браузер.

### Что ещё ломает наивный сбор

* **Отложенная подгрузка**: бесконечная прокрутка, кнопки «Показать ещё», ленивые изображения.
* **Оверлеи**: cookie-баннеры, модальные окна о регионе и подписке — они перекрывают клики.
* **Проверки браузера**: Cloudflare и подобные системы выполняют JS-challenge и проверяют отпечаток клиента. Их обход — не учебная задача: если площадка явно не хочет автоматических обращений, аналитик ищет официальный API или лицензированного поставщика данных.
* **Разметка без смысловых якорей**: сгенерированные классы вида `css-1x7ab3e`. Опора на такие классы — гарантированная поломка после следующего релиза.

### Экономика вопроса

Браузер дороже HTTP-запроса в десятки раз: запуск процесса, рендеринг, исполнение JS. Поэтому порядок выбора инструмента такой: **официальный API → скрытый JSON-эндпоинт → HTML без JS → headless-браузер**. Браузер — последний рубеж, а не первый.

## Теоретический блок 2. Selenium против Playwright

| Критерий | Selenium 4.x | Playwright 1.x |
|---|---|---|
| Протокол | W3C WebDriver: скрипт → драйвер → браузер | CDP (Chrome DevTools Protocol) напрямую, один WebSocket |
| Установка браузера | Chrome ставится отдельно, драйвер подбирает Selenium Manager | `playwright install chromium` кладёт совместимую сборку |
| Ожидания | неявные (`implicitly_wait`) и явные (`WebDriverWait` + `expected_conditions`) | автоожидание в каждом действии + явные `wait_for_selector` / `wait_for_function` |
| Асинхронность | синхронный API (есть сторонние обёртки) | синхронный и асинхронный API; в Jupyter обязателен асинхронный |
| Изоляция | новый профиль или новый браузер | `browser_context` — «инкогнито» за миллисекунды, десятки контекстов на один браузер |
| Сеть | перехват ограничен | `page.route()`: блокировка картинок, подмена ответов, чтение XHR |
| Скорость на типовом сценарии | ниже: лишний сетевой слой драйвера и ручные ожидания | выше: меньше слоёв, автоожидание убирает «сон на всякий случай» |
| Экосистема | огромная, много легаси-кода и примеров | моложе, документация лучше, API современнее |

**Практический вывод.** Для новых проектов сбора данных берут Playwright: меньше кода, меньше «плавающих» падений, дешевле параллелизм (контексты вместо браузеров). Selenium остаётся там, где он уже внедрён, нужен конкретный браузер или инфраструктура Selenium Grid. В этой работе основной путь — Playwright, а Selenium приведён как альтернативный движок: важно уметь читать и то, и другое.

## Теоретический блок 3. Шпаргалка по XPath

### Основы

| Выражение | Что делает |
|---|---|
| `//div` | все `div` на любом уровне |
| `//div/span` | `span` — прямой потомок `div` |
| `.//td` | потомки **текущего** узла (точка обязательна при поиске внутри строки) |
| `//*[@id="root"]` | элемент по атрибуту |
| `//td/@data-value` | значение атрибута |
| `//td/text()` | текстовый узел |
| `//tr[3]` | третья строка (нумерация с единицы) |
| `//tr[last()]`, `//tr[position() <= 5]` | последняя строка, первые пять |

### Функции, без которых не обойтись

| Функция | Пример | Смысл |
|---|---|---|
| `contains()` | `//div[contains(@class, "product")]` | класс среди нескольких: `class="card product new"` |
| `starts-with()` | `//tr[starts-with(@id, "row-")]` | префикс идентификатора |
| `text()` | `//th[text()="UPC"]` | точное совпадение текста узла |
| `normalize-space()` | `//td[normalize-space()="В наличии"]` | сравнение без лишних пробелов |
| `translate()` | `//button[contains(translate(., "ПРИНЯТЬ", "принять"), "принять")]` | сравнение без учёта регистра (в XPath 1.0 нет `lower-case()`) |
| `count()` | `count(//tr[@data-testid="asset-row"])` | сколько узлов — удобно в условии ожидания |

### Оси: движение по дереву

| Ось | Пример | Когда нужна |
|---|---|---|
| `following-sibling::` | `//th[text()="Объём"]/following-sibling::td[1]` | «ячейка справа от подписи» — таблицы характеристик |
| `preceding-sibling::` | `//td[@class="val"]/preceding-sibling::th` | найти подпись по значению |
| `parent::` / `..` | `//span[@class="price"]/parent::div` | подняться к карточке товара |
| `ancestor::` | `//a[@data-id="7"]/ancestor::tr` | строка таблицы, в которой лежит ссылка |

### Почему `data-testid` — лучший якорь

Классы вида `css-1x7ab3e` генерирует сборщик стилей: они меняются при каждом релизе. Атрибуты `data-testid`, `data-qa`, `data-cy` ставят разработчики для собственных автотестов, и ломать их себе дороже. Отсюда порядок надёжности селекторов:

```
data-testid  →  семантический атрибут (id, name, aria-label)  →  структура (tag + contains(@class))  →  текст  →  позиция
   надёжно                                                                                              хрупко
```

### Типичные ошибки

1. Забытая точка: внутри строки `//td` ищет по всему документу, а не в строке. Нужно `.//td`.
2. `text()` возвращает **список** текстовых узлов; если внутри есть вложенные теги, берите `string(.)` или `normalize-space(.)`.
3. Слишком длинный путь `/html/body/div[2]/div[3]/table/tbody/tr` — сломается от любой вставки блока.
4. `tbody`, которого нет в исходном HTML: браузер добавляет его сам, и XPath, скопированный из DevTools, не сработает на «сыром» HTML.

## Теоретический блок 4. Стратегии ожидания

Главная причина «плавающих» падений парсера — попытка прочитать элемент раньше, чем он появился.

| Стратегия | Как выглядит | Оценка |
|---|---|---|
| `time.sleep(5)` | «подождём на всякий случай» | Плохо: на быстрой сети теряем время, на медленной всё равно падаем |
| Неявное ожидание (`implicitly_wait`) | Selenium ждёт появления элемента до N секунд | Приемлемо как страховка, но не различает «нет элемента» и «элемент пустой» |
| Явное ожидание | `WebDriverWait(driver, 15).until(EC.presence_of_element_located(...))`, `page.wait_for_selector(...)` | Правильно: ждём конкретного условия |
| Ожидание предиката | `page.wait_for_function("() => document.querySelectorAll('tr').length > 12")` | Лучшее для догрузки: ждём **роста числа строк**, а не просто появления первой |
| Ожидание сети | `wait_until="networkidle"` | Осторожно: на страницах с постоянными опросами (poll) никогда не наступит |

### Важные различия

* `presence_of_element_located` — элемент есть в DOM; `visibility_of_element_located` — он ещё и виден. Кликать можно только по видимому и не перекрытому элементу — отсюда правило: **сначала закрыть баннер, потом кликать**.
* В Playwright автоожидание встроено в действия: `click()` сам дождётся видимости и стабильности элемента. Это не отменяет явных ожиданий там, где вы ждёте не элемент, а **состояние данных** (например, что строк стало больше).
* Любое ожидание конечно. Таймаут должен приводить не к падению всего ноутбука, а к понятному сообщению плюс сохранённым скриншоту и HTML — это артефакты для разбора.

### Шаг 2. Инициализация headless-браузера

**Методическое обоснование.** Браузер запускается один раз, а контекст — на каждую задачу: контекст даёт изоляцию cookie и кэша (аналог режима инкогнито) и стоит миллисекунды. Флаги `--no-sandbox` и `--disable-dev-shm-usage` обязательны в контейнере. Блокировка картинок и шрифтов через `page.route()` сокращает время загрузки в разы — на больших сборах это основной резерв скорости.

````python
#@title Шаг 2. Запуск headless-браузера { display-mode: "form" }
BLOCKED_TYPES = {"image", "font", "media"}

async def block_heavy(route, request):
    if request.resource_type in BLOCKED_TYPES:
        await route.abort()
    else:
        await route.continue_()

# ------------------------------------------------------------------ TODO 2.1
async def new_context(pw, headless=True):
    """Запустите Chromium и создайте контекст.

    Обязательно: аргументы --no-sandbox и --disable-dev-shm-usage (иначе Chromium не стартует
    в контейнере Colab), user_agent=UA, viewport 1440×900, locale="ru-RU",
    context.set_default_timeout(NAV_TIMEOUT_MS). При BLOCK_HEAVY_RESOURCES повесьте
    await context.route("**/*", block_heavy). Верните пару (browser, context).
    """
    # Запускаем Chromium
    browser = await pw.chromium.launch(
        headless=headless,
        args=[
            "--no-sandbox",
            "--disable-dev-shm-usage",
            "--disable-gpu",
            "--window-size=1440,900"
        ]
    )

    # Контекст
    context = await browser.new_context(
        user_agent=UA,
        viewport={"width": 1440, "height": 900},
        locale="ru-RU"
    )

    # Таймаут
    context.set_default_timeout(NAV_TIMEOUT_MS)

    # Блокируем тяжёлые ресурсы
    if BLOCK_HEAVY_RESOURCES:
        await context.route("**/*", block_heavy)

    return browser, context


async def smoke_test():
    async with async_playwright() as pw:
        browser, context = await new_context(pw, HEADLESS)
        page = await context.new_page()
        response = await page.goto(TARGET_URL, wait_until="domcontentloaded")
        info = {"версия браузера": browser.version, "код ответа": response.status if response else None,
                "заголовок страницы": await page.title()}
        await browser.close()
        return info

for k, v in (await smoke_test()).items():
    print(f"{k:>22}: {v}")
````

### Шаги 3–5. Баннеры, ожидания и сбор

**Методическое обоснование.** Три препятствия решаются тремя приёмами. Оверлей закрывается каскадом селекторов: сначала точный `data-testid`, затем текстовый поиск через `translate()` — и отсутствие баннера не считается ошибкой. Готовность данных проверяется явным ожиданием по числу строк, а не паузой. Догрузка выполняется циклом «клик → дождаться роста числа строк» с ограничением на число кликов, чтобы скрипт не зациклился.

Отдельный приём — **разделение труда**: браузер доводит страницу до нужного состояния и отдаёт финальный HTML, а разбор выполняется в `lxml`. Обращение к DOM через браузер стоит миллисекунды на каждый элемент; на тридцати строках по девять полей разница уже заметна, на тысячах — критична.

````python
#@title Шаги 3–5. Баннеры, явные ожидания и сбор через XPath { display-mode: "form" }
LOWER = "'АБВГДЕЁЖЗИЙКЛМНОПРСТУФХЦЧШЩЪЫЬЭЮЯABCDEFGHIJKLMNOPQRSTUVWXYZ'"
UPPER = "'абвгдеёжзийклмнопрстуфхцчшщъыьэюяabcdefghijklmnopqrstuvwxyz'"
OVERLAY_XPATHS = [
    "//button[@data-testid='cookie-accept']",
    f"//button[contains(translate(., {LOWER}, {UPPER}), 'принять')]",
    f"//button[contains(translate(., {LOWER}, {UPPER}), 'accept')]"
]

# ------------------------------------------------------------------ TODO 3.1
from time import perf_counter
async def dismiss_overlays(page, per_try_ms=1500):
    """Пройдите по OVERLAY_XPATHS, кликните по первому видимому элементу каждого типа.
    Важно: отсутствие баннера — штатная ситуация, функция не должна падать.
    Верните список закрытых селекторов."""

    closed = []

    for xpath in OVERLAY_XPATHS:
        try:
            locator = page.locator(f"xpath={xpath}")

            # Ограничиваем ожидание для каждого селектора
            await locator.first.wait_for(
                state="visible",
                timeout=per_try_ms
            )

            await locator.first.click(timeout=per_try_ms)
            closed.append(xpath)

            # Баннер закрыт - остальные селекторы проверять не нужно
            break

        except PWTimeout:
            continue

        except Exception as exc:
            # Если проблема с банером
            print(f"Не удалось обработать оверлей {xpath}: {exc}")

    return closed

# ------------------------------------------------------------------ TODO 4.1
async def wait_for_rows(page, row_xpath, min_rows=1, timeout=None):
    """Явное ожидание появления min_rows строк. Используйте page.wait_for_selector или
    page.wait_for_function с подсчётом узлов; time.sleep() — запрещён."""

    timeout = timeout or NAV_TIMEOUT_MS

    await page.wait_for_function(
        """
        ({xpath, minRows}) => {
            const result = document.evaluate(
                xpath,
                document,
                null,
                XPathResult.ORDERED_NODE_SNAPSHOT_TYPE,
                null
            );

            return result.snapshotLength >= minRows;
        }
        """,
        arg={
            "xpath": row_xpath,
            "minRows": min_rows
        },
        timeout=timeout
    )

    return await page.locator(f"xpath={row_xpath}").count()

# ------------------------------------------------------------------ TODO 5.1
async def load_all(page, row_xpath, more_xpath, max_clicks=12):
    """Кликайте «Показать ещё», пока кнопка видима и число строк растёт.
    После каждого клика — ожидание, а не пауза. Верните число кликов."""

async def load_all(page, row_xpath, more_xpath, max_clicks=12):
    """
    Нажимает кнопку «Показать ещё», пока появляются новые строки.
    После каждого клика ожидает увеличения количества строк.
    max_clicks защищает от бесконечного цикла.
    Возвращает число выполненных кликов.
    """
    clicks = 0
    rows = page.locator(f"xpath={row_xpath}")
    more = page.locator(f"xpath={more_xpath}")

    while clicks < max_clicks:
        if await more.count() == 0:
            break

        button = more.first

        if not await button.is_visible():
            break

        if not await button.is_enabled():
            break

        previous_count = await rows.count()

        try:
            await button.click()
            clicks += 1

            # Ждём именно увеличения числа строк, а не просто появления кнопки.
            await page.wait_for_function(
                """
                ({xpath, previousCount}) => {
                    const result = document.evaluate(
                        xpath,
                        document,
                        null,
                        XPathResult.ORDERED_NODE_SNAPSHOT_TYPE,
                        null
                    );

                    return result.snapshotLength > previousCount;
                }
                """,
                arg={
                    "xpath": row_xpath,
                    "previousCount": previous_count
                },
                timeout=NAV_TIMEOUT_MS
            )

        except PWTimeout:
            # Новых строк нет: достигнут конец данных либо кнопка
            # больше не загружает записи.
            print("Количество строк не увеличилось. Завершаем догрузку.")
            break

    print(f"Кликов по кнопке догрузки: {clicks}")
    print(f"Всего строк после догрузки: {await rows.count()}")

    return clicks


# ------------------------------------------------------------------ TODO 5.2
async def collect(url, cfg):
    """Сценарий целиком: goto → dismiss_overlays → wait_for_rows → load_all →
    page.content(). При таймауте сохраните скриншот и HTML в WORK_DIR и сообщите об этом
    понятным текстом. Верните (html, образец первой строки, время на locator)."""

    work_dir = Path(WORK_DIR)
    work_dir.mkdir(parents=True, exist_ok=True)

    browser = None
    context = None
    page = None

    try:
        async with async_playwright() as pw:
            browser, context = await new_context(pw, HEADLESS)
            page = await context.new_page()

            response = await page.goto(
                url,
                wait_until="domcontentloaded",
                timeout=NAV_TIMEOUT_MS
            )

            if response and response.status >= 400:
                raise RuntimeError(
                    f"Страница вернула HTTP {response.status}: {url}"
                )

            # Шаг 3: обработка оверлеев.
            closed = await dismiss_overlays(page)
            print(f"Закрытые оверлеи: {len(closed)}")

            # Шаг 4: ожидание отрисовки данных.
            await wait_for_rows(
                page,
                cfg["row_xpath"],
                min_rows=1,
                timeout=NAV_TIMEOUT_MS
            )

            # Шаг 5: догрузка оставшихся записей.
            await load_all(
                page,
                cfg["row_xpath"],
                cfg["more_xpath"],
                max_clicks=12
            )

            # Получаем HTML после завершения динамической подгрузки.
            html_text = await page.content()

            # Образец первой строки для проверки структуры.
            rows = page.locator(f"xpath={cfg['row_xpath']}")
            sample = await rows.first.inner_text()

            # Измеряем время обращения к locator.
            start = perf_counter()
            locator_count = await rows.count()
            t_locator = (perf_counter() - start) * 1000

            # Сохраняем артефакты успешного сбора.
            (work_dir / "market_monitor.html").write_text(
                html_text,
                encoding="utf-8"
            )

            await page.screenshot(
                path=str(work_dir / "page.png"),
                full_page=True
            )

            print(f"Строк по locator: {locator_count}")
            print(f"Время locator: {t_locator:.3f} мс")
            print(f"HTML сохранён: {work_dir / 'market_monitor.html'}")
            print(f"Скриншот сохранён: {work_dir / 'page.png'}")

            await browser.close()

            return html_text, sample, t_locator

    except PWTimeout as exc:
        # При таймауте сохраняем состояние страницы для диагностики.
        if page is not None:
            try:
                await page.screenshot(
                    path=str(work_dir / "timeout.png"),
                    full_page=True
                )
                (work_dir / "timeout.html").write_text(
                    await page.content(),
                    encoding="utf-8"
                )
            except Exception:
                pass

        raise RuntimeError(
            "Истекло время ожидания страницы или динамических данных. "
            f"Диагностические файлы сохранены в {work_dir}."
        ) from exc

    finally:
        if context is not None:
            try:
                await context.close()
            except Exception:
                pass

        if browser is not None:
            try:
                await browser.close()
            except Exception:
                pass


# ------------------------------------------------------------------ TODO 5.3
def parse_html(html, cfg):
    """Разберите сохранённый HTML через lxml тем же XPath: список словарей по полям cfg["fields"]."""
    if not html or not html.strip():
        raise ValueError("Получен пустой HTML для разбора.")

    tree = lxml_html.fromstring(html)

    rows = tree.xpath(cfg["row_xpath"])
    records = []

    for row in rows:
        record = {}

        for field in cfg["fields"]:
            # В стенде поля таблицы размечены через data-testid.
            # Точка в XPath ограничивает поиск текущей строкой.
            nodes = row.xpath(
                f".//*[@data-testid='{field}']"
            )

            if nodes:
                value = " ".join(
                    "".join(nodes[0].itertext()).split()
                )
            else:
                value = None

            record[field] = value

        records.append(record)

    return records

html, sample, t_locator = await collect(TARGET_URL, CFG)
records = parse_html(html, CFG)
raw_df = pd.DataFrame(records)
print(f"Собрано строк: {len(raw_df)}")
display(raw_df.head(8))
````

**Интерпретация.** Смотрите на четыре числа. Сколько оверлеев закрыто — если ноль, а баннер был, клики не сработали и дальше будет перекрытие. Сколько строк после гидратации — это размер первой «порции». Сколько кликов «показать ещё» и сколько строк получилось в итоге — при правильной догрузке итог равен полному размеру витрины (на стенде 30). И сравнение времени: один элемент через браузер против всей таблицы через `lxml` — именно эта разница объясняет, почему промышленные парсеры забирают HTML целиком.

### Шаг 6. Очистка признаков и выгрузка

**Методическое обоснование.** С витрины приходят строки в человеческом формате: `30.5M`, `1.23B`, `+2,5%`, `12 345,60 ₽`, `n/a`, `—`. Каждая из этих форм — отдельная ловушка: неразрывный пробел, запятая как десятичный разделитель, Unicode-минус вместо дефиса, пропуск, замаскированный под текст. Функции очистки пишутся с тестами и только потом применяются к данным. Excel выбран как формат обмена с бизнесом: с числовыми форматами, закреплённой шапкой и отдельным листом агрегатов.

````python
#@title Шаг 6. Очистка признаков и выгрузка в Excel { display-mode: "form" }
NBSP = "\u00a0\u202f\u2009"
MINUS = "\u2212\u2013\u2014"     # на сайтах часто стоит Unicode-минус, а не дефис
SUFFIXES = {"k": 1e3, "к": 1e3, "тыс": 1e3, "m": 1e6, "м": 1e6, "млн": 1e6,
            "b": 1e9, "млрд": 1e9, "t": 1e12, "трлн": 1e12}
MISSING = {"", "n/a", "na", "—", "-", "–", "нет данных", "none", "null"}

def _norm_number(raw):
    """'1 234,5' → 1234.5 | '1,234.56' → 1234.56. Разделитель определяем по последнему знаку."""
    s = re.sub(rf"[{NBSP} ]", "", raw)
    if "," in s and "." in s:
        s = s.replace(",", "") if s.rfind(".") > s.rfind(",") else s.replace(".", "").replace(",", ".")
    elif "," in s:
        s = s.replace(",", ".") if len(s.split(",")[-1]) <= 2 else s.replace(",", "")
    return float(s)

# ------------------------------------------------------------------ TODO 6.1
# Регулярное выражение для чисел с необязательным знаком и суффиксом
NUM_RE = re.compile(
    r"(?P<sign>[-+\u2212\u2013\u2014])?"
    r"(?P<num>\d[\d\s\u00a0\u202f\u2009,\.]*\d|\d)"
    r"\s*"
    r"(?P<suffix>млрд|млн|тыс|[kкmмbt])?",
    re.IGNORECASE
)

# Регулярное выражение для распознавания валют
CUR_RE = re.compile(
    r"(?<![A-Za-z])(?:USD|EUR|RUB|руб\.?|usd|eur|rub)|[$€£₽]",
    re.IGNORECASE
)

def parse_suffix_number(text):
    r"""'30.5M' → 30500000.0 | '1.23B' → 1230000000.0 | '1 234,5' → 1234.5 | 'n/a' → None
    Подсказка: regex (?P<sign>[-+\u2212])?(?P<num>\d[\d\s,.]*)(?P<suffix>млрд|млн|тыс|[kкmмbt])?,
    затем убрать пробелы, определить десятичный разделитель и умножить на множитель суффикса."""
    if text is None or str(text).strip().lower() in MISSING:
        return None
    m = NUM_RE.search(str(text))
    if not m or not m.group("num"):
        return None
    try:
        value = _norm_number(m.group("num"))
    except ValueError:
        return None
    if m.group("suffix"):
        value *= SUFFIXES[m.group("suffix").lower()]
    if m.group("sign") and m.group("sign") in "-" + MINUS:
        value = -value
    return value

# ------------------------------------------------------------------ TODO 6.2
def parse_percent(text):
    """'+2.5%' → 0.025 | '−1,8%' → -0.018 | '—' → None. Проценты храним долями единицы."""
    if text is None or str(text).strip().lower() in MISSING:
        return None
    value = parse_suffix_number(str(text).replace("%", ""))
    return None if value is None else round(value / 100, 6)

CUR_MAP = {
    "$": "USD",
    "€": "EUR",
    "£": "GBP",
    "₽": "RUB",
    "р": "RUB",
    "руб": "RUB",
    "руб.": "RUB",
    "usd": "USD",
    "eur": "EUR",
    "rub": "RUB",
}

# ------------------------------------------------------------------ TODO 6.3
def parse_money(text):
    """'$1,234.56' → (1234.56, 'USD') | '12 345,60 ₽' → (12345.6, 'RUB')"""
    if text is None or str(text).strip().lower() in MISSING:
        return (None, None)

    s = str(text)
    m = CUR_RE.search(s)

    if m:
        token = m.group(0).lower()
        cur = CUR_MAP.get(token)
        s = CUR_RE.sub("", s)
    else:
        cur = None

    return (parse_suffix_number(s), cur)
# ------------------------------------------------------------------ TODO 6.4
def parse_date_ru(text):
    """'18.09.2021' и '2021-09-18' → pd.Timestamp, иначе pd.NaT"""
    if text is None or pd.isna(text):
         return pd.NaT

    s = str(text or "").strip()
    if not s:
        return pd.NaT

    for fmt in ("%d.%m.%Y", "%Y-%m-%d", "%d/%m/%Y"):
        try:
            return pd.Timestamp(pd.to_datetime(s, format=fmt, errors="raise"))
        except (ValueError, TypeError):
            continue

    return pd.NaT



CASES = [
    (parse_suffix_number, "30.5M", 30_500_000.0), (parse_suffix_number, "1.23B", 1_230_000_000.0),
    (parse_suffix_number, "870K", 870_000.0), (parse_suffix_number, "1 234,5", 1234.5),
    (parse_suffix_number, "n/a", None),
    (parse_percent, "+2.5%", 0.025), (parse_percent, "\u22121,80%", -0.018), (parse_percent, "—", None),
    (parse_money, "$1,234.56", (1234.56, "USD")), (parse_money, "12 345,60 ₽", (12345.6, "RUB")),
    (parse_date_ru, "18.09.2021", pd.Timestamp("2021-09-18")),
]
for func, arg, expected in CASES:
    got = func(arg)
    assert got == expected, f"{func.__name__}({arg!r}) → {got!r}, ожидалось {expected!r}"
print("Функции очистки прошли проверку")

# ------------------------------------------------------------------ TODO 6.5
# Соберите df: числовые колонки price_value, change_24h, volume_value, mcap_value, listed_date,
# посчитайте долю распознанных значений по каждой колонке и выгрузите результат в Excel
# (лист data + лист by_sector, закреплённая шапка, числовые форматы, ширина колонок).


# ------------------------------------------------------------------
# TODO 6.5. Очистка DataFrame, оценка качества данных и экспорт в Excel
from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.utils import get_column_letter
df = raw_df.copy()

# Преобразуем признаки в числовые значения и даты
df["price_value"] = df["price"].apply(
    lambda x: parse_money(x)[0]
)
# Решила ещё добавить валюту как в примере Тимура Муртазовича на паре
df["currency"] = df["price"].apply(
    lambda x: parse_money(x)[1]
)

df["change_24h"] = df["change"].apply(
    parse_percent
)

df["volume_value"] = df["volume"].apply(
    parse_suffix_number
)

df["mcap_value"] = df["mcap"].apply(
    parse_suffix_number
)

df["listed_date"] = df["listed"].apply(
    parse_date_ru
)

# Рассчитываем долю распознанных значений
clean_cols = [
    "price_value",
    "change_24h",
    "volume_value",
    "mcap_value",
    "listed_date",
]

column_names_ru = {
    "price_value": "Цена",
    "change_24h": "Изменение за 24 ч.",
    "volume_value": "Объём торгов",
    "mcap_value": "Капитализация",
    "listed_date": "Дата листинга",
}

quality_df = pd.DataFrame({
    "Столбец": [column_names_ru[col] for col in clean_cols],
    "Распознано": [df[col].notna().sum() for col in clean_cols],
    "Всего": [len(df) for col in clean_cols],
    "Доля, %": [
        round(df[col].notna().mean() * 100, 2)
        for col in clean_cols
    ],
})

print("Качество распознавания данных:")
display(quality_df)

# Удаляем исходные столбцы с неочищенными значениями, чтобы таблица не была слишком большой
raw_cols = ["price", "change", "volume", "mcap", "listed"]
df = df.drop(columns=raw_cols, errors="ignore")

# Создаём сводку по секторам
by_sector = (
    df.groupby("sector", dropna=False)
    .agg(
        price_mean=("price_value", "mean"),
        change_mean=("change_24h", "mean"),
        volume_mean=("volume_value", "mean"),
        mcap_mean=("mcap_value", "mean"),
    )
    .reset_index()
)

print("Сводка по секторам создана:")
display(by_sector.head())

# Выбираем папку и имя файла

EXCEL_PATH = Path("/content/lr1/market_monitor_clean.xlsx")

# Выгружаем данные в Excel
with pd.ExcelWriter(
    EXCEL_PATH,
    engine="openpyxl",
    date_format="DD.MM.YYYY",
    datetime_format="DD.MM.YYYY",
) as writer:
    df.to_excel(
        writer,
        sheet_name="data",
        index=False,
    )

    by_sector.to_excel(
        writer,
        sheet_name="by_sector",
        index=False,
    )

    # Форматируем оба листа
    for sheet_name in ["data", "by_sector"]:
        ws = writer.sheets[sheet_name]

        # Закрепляем заголовок
        ws.freeze_panes = "A2"

        # Включаем автофильтр
        ws.auto_filter.ref = ws.dimensions

        # Оформляем заголовки и ширину столбцов
        for cell in ws[1]:
            cell.font = Font(bold=True, color="FFFFFF")
            cell.fill = PatternFill(
                fill_type="solid",
                fgColor="1F4E78",
            )
            cell.alignment = Alignment(
                horizontal="center",
                vertical="center",
            )

        ws.row_dimensions[1].height = 25

        for col_idx in range(1, ws.max_column + 1):
            letter = get_column_letter(col_idx)

            max_len = max(
                len(str(ws.cell(row=r, column=col_idx).value or ""))
                for r in range(1, ws.max_row + 1)
            )

            ws.column_dimensions[letter].width = min(
                max(max_len + 2, 12),
                30,
            )

    # Числовые форматы на листе data
    ws = writer.sheets["data"]
    data_cols = {
        "price_value": "#,##0.00",
        "change_24h": "0.00%",
        "volume_value": "#,##0",
        "mcap_value": "#,##0",
        "listed_date": "DD.MM.YYYY",
    }

    data_headers = {
        cell.value: cell.column
        for cell in ws[1]
    }

    for col_name, number_format in data_cols.items():
        col_idx = data_headers[col_name]

        for row_idx in range(2, ws.max_row + 1):
            ws.cell(
                row=row_idx,
                column=col_idx,
            ).number_format = number_format

    # Числовые форматы на листе by_sector
    ws = writer.sheets["by_sector"]

    sector_formats = {
        "price_mean": "#,##0.00",
        "change_mean": "0.00%",
        "volume_mean": "#,##0",
        "mcap_mean": "#,##0",
    }

    sector_headers = {
        cell.value: cell.column
        for cell in ws[1]
    }

    for col_name, number_format in sector_formats.items():
        col_idx = sector_headers[col_name]

        for row_idx in range(2, ws.max_row + 1):
            ws.cell(
                row=row_idx,
                column=col_idx,
            ).number_format = number_format

print(f"Excel-файл сохранён: {EXCEL_PATH}")
print(f"Размер исходной таблицы: {df.shape}")
print(f"Размер сводки по секторам: {by_sector.shape}")

print('\nОбщая таблица после преобразования:')
display(df.head(10))
print('\n\n\nСводка по секторам:')
display(by_sector)
````

**Интерпретация.** Сначала блок тестов: пока он не сообщает «все тесты пройдены», к данным переходить рано. Затем таблица качества очистки — доля распознанных значений по каждой колонке. Обратите внимание, что цена и изменение распознаются не на 100 %: на стенде намеренно оставлены `n/a` и `—`. Это правильное поведение — пропуск должен остаться пропуском, а не превратиться в ноль, иначе он занизит средние. Файл Excel открывается в бизнес-подразделении, поэтому проценты сохранены как доли с процентным форматом, а не как «2,5».

### Шаг 7. Бизнес-анализ и визуализация

**Методическое обоснование.** Сбор данных — не результат, а полуфабрикат. Результат — решение. Поэтому строим четыре представления: топ по обороту (где ликвидность), распределение изменения (каков режим рынка), когорты лидеров роста и падения (что выбивается) и связь капитализации с оборотом (насколько крупные позиции реально торгуются). Итог — короткий текст для ЛПР, в котором каждое утверждение опирается на посчитанное число.

````python
#@title Шаг 7. Бизнес-анализ и визуализация { display-mode: "form" }
# ------------------------------------------------------------------ TODO 7.1
# Постройте четыре графика (см. столбец «Аналитическая задача» вашего варианта):
#   1) топ-10 по ключевой метрике с выделением лидера отдельным цветом;
#   2) распределение относительного изменения с опорной линией нуля и медианы;
#   3) когорты лидеров роста и падения (по 5 позиций);
#   4) связь двух количественных признаков (при необходимости — логарифмические шкалы).
# Требования: заголовок с объёмом выборки, подписи осей с единицами измерения, читаемый шрифт,
# единая палитра, отсутствие «мусорных» подписей.

PERIOD_LABEL = "срез на дату выгрузки"

# Единая палитра
COLORS = {
    "accent": "#2878B5",    # основной акцент
    "neutral": "#C8D2DC",   # остальные объекты
    "text": "#263238",
    "grid": "#E6EBF0",
    "growth": "#238636",    # рост
    "fall": "#D64545",      # падение
    "median": "#E09F3E",
}

plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "font.size": 10,
    "axes.titlesize": 12,
    "axes.labelsize": 10,
    "xtick.labelsize": 9,
    "ytick.labelsize": 9,
    "figure.dpi": 120,
    "savefig.dpi": 160,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

# Работаем с копией, чтобы не менять исходный df
plot_df = df.copy()

required_cols = [
    "change_24h",
    "volume_value",
    "mcap_value",
]

missing_cols = [c for c in required_cols if c not in plot_df.columns]
if missing_cols:
    raise ValueError(f"В df отсутствуют необходимые колонки: {missing_cols}")

# Выбираем подпись актива автоматически
label_candidates = [
    "name", "coin", "asset", "symbol", "ticker",
    "Название", "Монета", "Актив", "Тикер"
]
label_col = next(
    (c for c in label_candidates if c in plot_df.columns),
    None
)

# Добавляем номер к повторяющимся названиям,
# чтобы подписи на графиках не путались.
plot_df["_asset_label"] = plot_df[label_col]
dup_mask = plot_df["_asset_label"].duplicated(keep=False)
if dup_mask.any():
    plot_df.loc[dup_mask, "_asset_label"] = (
        plot_df.loc[dup_mask, "_asset_label"]
        + " ["
        + plot_df.loc[dup_mask].groupby("_asset_label").cumcount().add(1).astype(str)
        + "]"
    )

# Числовые колонки: приводим к числовому типу
for col in required_cols:
    plot_df[col] = pd.to_numeric(plot_df[col], errors="coerce")

# Единицы измерения (Перед построением графика с единицами измерения сделала проверку - если разные, то строиться не будет, т.к. неразумно сравнивать в рублях и долларах, например)
currency_value = (
    df["currency"]
    .dropna()
    .astype(str)
    .str.strip()
)
currency_value = currency_value[currency_value != ""]

if currency_value.empty:
    raise ValueError("В колонке 'currency' нет заполненных значений.")

CURRENCY = currency_value.iloc[0]

VALUE_UNIT = f"млн {CURRENCY}"
VALUE_DIVISOR = 1_000_000

plot_df["_volume_mln"] = plot_df["volume_value"] / VALUE_DIVISOR
plot_df["_mcap_mln"] = plot_df["mcap_value"] / VALUE_DIVISOR

N_TOTAL = len(plot_df)

def make_title(title, n):
    return f"{title}\nn = {n:,} | {PERIOD_LABEL}".replace(",", " ")

def style_ax(ax):
    ax.grid(axis="both", color=COLORS["grid"], linewidth=0.8)
    ax.set_axisbelow(True)
    ax.tick_params(axis="both", labelsize=9)
    for spine in ["left", "bottom"]:
        ax.spines[spine].set_color(COLORS["grid"])

output_dir = Path("/content/lr1")
output_dir.mkdir(parents=True, exist_ok=True)

def save_fig(fig, filename):
    fig.tight_layout()
    fig.savefig(output_dir / filename, bbox_inches="tight")
    plt.show()

# График 1. Доля отсутствующих значений по признакам

quality_cols = {
    "Цена": "price_value",
    "Изменение за 24 ч.": "change_24h",
    "Объём торгов": "volume_value",
    "Капитализация": "mcap_value",
    "Дата листинга": "listed_date",
}

quality_rows = []
for label, col in quality_cols.items():
    if col in plot_df.columns:
        missing_count = int(plot_df[col].isna().sum())
        quality_rows.append({
            "Признак": label,
            "Пропуски, %": (
                100 * missing_count / N_TOTAL if N_TOTAL else np.nan
            ),
            "Пропущено, шт.": missing_count,
        })

quality_plot = pd.DataFrame(quality_rows).sort_values(
    "Пропуски, %", ascending=True
)

fig, ax = plt.subplots(figsize=(10, 5.5))
bars = ax.barh(
    quality_plot["Признак"],
    quality_plot["Пропуски, %"],
    color=COLORS["neutral"],
    edgecolor="white",
)

# Выделяем признак с максимальной долей пропусков (если максимумов несколько, то выделяю все из них)
if len(bars):
    max_value = quality_plot["Пропуски, %"].max()

    for i, value in enumerate(quality_plot["Пропуски, %"]):
        if value == max_value:
            bars[i].set_color(COLORS["accent"])

for bar, (_, row) in zip(bars, quality_plot.iterrows()):
    ax.text(
        bar.get_width() + 0.5,
        bar.get_y() + bar.get_height() / 2,
        f'{row["Пропуски, %"]:.1f}% ({row["Пропущено, шт."]} шт.)',
        va="center",
        fontsize=9,
    )

ax.set_xlim(0, max(105, quality_plot["Пропуски, %"].max() * 1.2))
ax.set_xlabel("Доля отсутствующих значений, %")
ax.set_ylabel("Признак")
ax.set_title(make_title("Полнота данных по признакам", N_TOTAL))
style_ax(ax)
save_fig(fig, "01_missing_values.png")

# График 2. Доля отсутствующих значений (круговая)

quality_cols = [
    "price_value",
    "change_24h",
    "volume_value",
    "mcap_value",
    "listed_date",
]

available_cols = [c for c in quality_cols if c in df.columns]

total_cells = len(df) * len(available_cols)
missing_cells = int(df[available_cols].isna().sum().sum())
filled_cells = total_cells - missing_cells

if total_cells > 0:
    values = [filled_cells, missing_cells]
    labels = [
        f"Заполнено\n{filled_cells:,} ячеек".replace(",", " "),
        f"Пропуски\n{missing_cells:,} ячеек".replace(",", " "),
    ]

    fig, ax = plt.subplots(figsize=(10, 6.5))

    wedges, _ = ax.pie(
        values,
        colors=["#2878B5", "#C8D2DC"],
        startangle=90,
        counterclock=False,
        wedgeprops={"edgecolor": "white", "linewidth": 2},
    )

    percentages = [
        value / total_cells * 100
        for value in values
    ]

    legend_labels = [
        f"{labels[i]} — {percentages[i]:.1f}%"
        for i in range(2)
    ]

    ax.legend(
        wedges,
        legend_labels,
        loc="lower center",
        bbox_to_anchor=(0.5, -0.08),
        frameon=False,
        fontsize=9,
    )

    ax.set_title(
        f"Полнота данных\nn = {len(df)} объектов | "
        f"{len(available_cols)} признаков"
    )

    save_fig(fig, "02_data_completeness_pie.png")

# График 3. Топ-10 по объёму торгов

# График будет корректным только в случае, если у нас одна валюта. Поэтому добавляем проверку


# График 3. Топ-10 по объёму торгов
# Строим график только при единой валюте среди заполненных значений currency

if "currency" not in plot_df.columns:
    print("График 3 не построен: отсутствует колонка 'currency'.")
else:
    # Исключаем пропуски и пустые строки
    currency_values = (
        plot_df["currency"]
        .dropna()
        .astype(str)
        .str.strip()
    )
    currency_values = currency_values[currency_values != ""]

    unique_currencies = currency_values.unique()

    if len(unique_currencies) == 1:
        # Валюта единая - можно строить график
        currency = unique_currencies[0]

        volume_data = plot_df.dropna(subset=["volume_value"]).copy()
        volume_data = volume_data[volume_data["volume_value"] >= 0]

        top10 = volume_data.nlargest(10, "volume_value").sort_values(
            "volume_value", ascending=True
        )

        fig, ax = plt.subplots(figsize=(10, 6))

        bar_colors = [COLORS["neutral"]] * len(top10)
        if len(top10):
            bar_colors[-1] = COLORS["accent"]

        bars = ax.barh(
            top10["_asset_label"],
            top10["_volume_mln"],
            color=bar_colors,
        )

        for bar, value in zip(bars, top10["_volume_mln"]):
            ax.text(
                bar.get_width(),
                bar.get_y() + bar.get_height() / 2,
                f" {value:,.2f}".replace(",", " "),
                va="center",
                fontsize=9,
            )

        ax.set_xlabel(f"Объём торгов, млн {currency}")
        ax.set_ylabel("Актив")
        ax.set_title(
            make_title(
                f"Топ-10 активов по объёму торгов ({currency})",
                len(volume_data),
            )
        )

        style_ax(ax)
        save_fig(fig, "03_top10_volume.png")

# График 4. Распределение изменения за 24 часа

changes = (
    plot_df["change_24h"].dropna().replace([np.inf, -np.inf], np.nan).dropna()
    * 100
)

fig, ax = plt.subplots(figsize=(10, 5.5))

if len(changes):
    ax.hist(
        changes,
        bins="auto",
        color=COLORS["neutral"],
        edgecolor="white",
    )
    median_change = changes.median()

    ax.axvline(
        0,
        color=COLORS["text"],
        linewidth=1.5,
        linestyle="--",
        label="Нулевое изменение",
    )
    ax.axvline(
        median_change,
        color=COLORS["median"],
        linewidth=2,
        label=f"Медиана: {median_change:.2f}%",
    )
    ax.legend(fontsize=9, frameon=False)
else:
    ax.text(
        0.5, 0.5, "Нет данных для построения",
        transform=ax.transAxes, ha="center"
    )

ax.set_xlabel("Изменение за 24 часа, %")
ax.set_ylabel("Число активов, шт.")
ax.set_title(make_title("Распределение изменения за 24 часа", len(changes)))
style_ax(ax)
save_fig(fig, "04_change_distribution.png")

# График 5. Пять лидеров роста

growth_data = plot_df.dropna(subset=["change_24h"]).copy()
growth_data = growth_data[
    np.isfinite(growth_data["change_24h"])
    & (growth_data["change_24h"] > 0)
]
top_growth = growth_data.nlargest(5, "change_24h").sort_values(
    "change_24h", ascending=True
)

fig, ax = plt.subplots(figsize=(10, 4.5))

if len(top_growth):
    bars = ax.barh(
        top_growth["_asset_label"],
        top_growth["change_24h"] * 100,
        color=COLORS["growth"],
    )
    for bar, value in zip(bars, top_growth["change_24h"] * 100):
        ax.text(
            bar.get_width(),
            bar.get_y() + bar.get_height() / 2,
            f" +{value:.2f}%",
            va="center",
            fontsize=9,
        )
    ax.axvline(0, color=COLORS["text"], linewidth=1)
else:
    ax.text(
        0.5, 0.5, "Положительных изменений нет",
        transform=ax.transAxes, ha="center"
    )

ax.set_xlabel("Изменение за 24 часа, %")
ax.set_ylabel("Актив")
ax.set_title(make_title("Лидеры роста: топ-5", len(growth_data)))
style_ax(ax)
save_fig(fig, "05_top5_growth.png")

# График 6. Пять лидеров падения

fall_data = plot_df.dropna(subset=["change_24h"]).copy()
fall_data = fall_data[
    np.isfinite(fall_data["change_24h"])
    & (fall_data["change_24h"] < 0)
]
top_fall = fall_data.nsmallest(5, "change_24h").sort_values(
    "change_24h", ascending=False
)

fig, ax = plt.subplots(figsize=(10, 4.5))

if len(top_fall):
    bars = ax.barh(
        top_fall["_asset_label"],
        top_fall["change_24h"] * 100,
        color=COLORS["fall"],
    )
    for bar, value in zip(bars, top_fall["change_24h"] * 100):
        ax.text(
            bar.get_width(),
            bar.get_y() + bar.get_height() / 2,
            f" {value:.2f}%",
            va="center",
            ha="right",
            fontsize=9,
            color="white",
        )
    ax.axvline(0, color=COLORS["text"], linewidth=1)
else:
    ax.text(
        0.5, 0.5, "Отрицательных изменений нет",
        transform=ax.transAxes, ha="center"
    )

ax.set_xlabel("Изменение за 24 часа, %")
ax.set_ylabel("Актив")
ax.set_title(make_title("Лидеры падения: топ-5", len(fall_data)))
style_ax(ax)
save_fig(fig, "06_top5_fall.png")

PLOTS_DONE = True

# ------------------------------------------------------------------ TODO 7.2
# Сформируйте текстовый отчёт для ЛПР: что собрано, три наблюдения с числами,
# два-три управленческих решения с ожидаемым эффектом и способом проверки.

from datetime import datetime
from pathlib import Path

REPORT_DATE = datetime.now().strftime("%d.%m.%Y")

report_df = df.copy()
n_total = len(report_df)

# Приведение числовых показателей к числовому типу
numeric_candidates = [
    "price_value", "volume_value", "mcap_value", "change_24h"
]

for col in numeric_candidates:
    if col in report_df.columns:
        report_df[col] = pd.to_numeric(
            report_df[col], errors="coerce"
        )

def pct(part, total):
    #Доля в процентах.
    return 100 * part / total if total else 0.0

def fmt_num(value, digits=2):
    #Форматирование числа для русского отчёта.
    if pd.isna(value):
        return "нет данных"
    return f"{value:,.{digits}f}".replace(",", " ").replace(".", ",")

def find_column(candidates):
    #оиск колонки по нескольким возможным названиям.
    return next(
        (c for c in candidates if c in report_df.columns),
        None
    )

# Название товара
name_col = find_column([
    "name", "Название", "product_name", "title",
    "ticker", "symbol", "asset", "coin"
])

# Категория товара
category_col = find_column([
    "category", "Категория", "sector", "category_name"
])

# Цена
price_col = find_column([
    "price_value", "price", "Цена", "product_price"
])

# Признаки фактического наличия товара
stock_col = find_column([
    "stock", "stock_quantity", "quantity",
    "остаток", "остаток_шт", "available_quantity",
    "inventory", "inventory_quantity"
])

status_col = find_column([
    "availability", "stock_status", "status",
    "наличие", "статус", "availability_status"
])

# 1. Полнота данных
quality_cols = [
    c for c in [
        price_col, "change_24h", "volume_value",
        "mcap_value", "listed_date"
    ]
    if c is not None and c in report_df.columns
]
quality_cols = list(dict.fromkeys(quality_cols))

if n_total:
    missing_price = (
        int(report_df[price_col].isna().sum())
        if price_col else None
    )
    price_missing_pct = (
        pct(missing_price, n_total)
        if missing_price is not None else None
    )

    if quality_cols:
        total_cells = n_total * len(quality_cols)
        missing_cells = int(
            report_df[quality_cols].isna().sum().sum()
        )
        filled_cells = total_cells - missing_cells
        completeness = pct(filled_cells, total_cells)
    else:
        missing_cells = None
        filled_cells = None
        completeness = None
else:
    missing_price = None
    price_missing_pct = None
    missing_cells = None
    filled_cells = None
    completeness = None

# 2. Фактический дефицит: только при наличии соответствующего поля
shortage_count = None
shortage_pct = None
shortage_basis = None

if stock_col:
    stock = pd.to_numeric(
        report_df[stock_col], errors="coerce"
    )
    known = stock.notna()
    shortage_count = int((stock[known] <= 0).sum())
    shortage_pct = pct(shortage_count, int(known.sum()))
    shortage_basis = (
        f"остаток товара ({stock_col}); "
        f"проверено {int(known.sum())} позиций"
    )

elif status_col:
    status = (
        report_df[status_col]
        .fillna("")
        .astype(str)
        .str.strip()
        .str.lower()
    )
    known_labels = [
        "в наличии", "есть в наличии", "нет в наличии",
        "нет", "out of stock", "in stock",
        "outofstock", "instock", "unavailable", "available"
    ]
    known = status.isin(known_labels)

    shortage_labels = [
        "нет в наличии", "нет", "out of stock",
        "outofstock", "unavailable"
    ]

    shortage_count = int(status[known].isin(shortage_labels).sum())
    shortage_pct = pct(shortage_count, int(known.sum()))
    shortage_basis = (
        f"статус наличия ({status_col}); "
        f"проверено {int(known.sum())} позиций"
    )

# 3. Дополнительные числовые наблюдения
change_col = find_column(["change_24h", "change", "Изменение"])
change_median = None
positive_count = None
negative_count = None
change_n = 0

if change_col:
    changes = pd.to_numeric(
        report_df[change_col], errors="coerce"
    ).replace([np.inf, -np.inf], np.nan).dropna()

    change_n = len(changes)

    if change_n:
        # change_24h в исходном шаблоне хранится как доля,
        # например 0.05 = 5%. change в процентах не умножаем.
        multiplier = 100 if change_col == "change_24h" else 1
        changes_pct = changes * multiplier

        change_median = float(changes_pct.median())
        positive_count = int((changes_pct > 0).sum())
        negative_count = int((changes_pct < 0).sum())

# 4. Топ категорий по пропускам цены
category_observation = None

if category_col and price_col and n_total:
    temp = report_df[[category_col, price_col]].copy()
    temp[category_col] = (
        temp[category_col].fillna("Не указана").astype(str)
    )
    grouped = temp.groupby(category_col).agg(
        positions=(price_col, "size"),
        missing_prices=(price_col, lambda s: int(s.isna().sum()))
    )
    grouped["missing_pct"] = (
        100 * grouped["missing_prices"] / grouped["positions"]
    )

    # Не рассматриваем категории из одной позиции
    eligible = grouped[grouped["positions"] >= 2]

    if not eligible.empty:
        eligible = eligible.sort_values(
            ["missing_pct", "missing_prices"],
            ascending=False
        )

        worst_category = eligible.index[0]
        worst = eligible.iloc[0]

        category_observation = (
            f"Категория «{worst_category}»: "
            f"{int(worst['missing_prices'])} пропусков цены "
            f"из {int(worst['positions'])} позиций "
            f"({fmt_num(worst['missing_pct'], 1)}%)."
        )

# Т.к. требуется мониторинг, то часть метрик и наблюдений формируется автоматически. А вот управленческие решения задаются вручную.
# 5. Три наблюдения с числами

def find_market_col(candidates):
    """Находит первый существующий столбец из списка."""
    for col in candidates:
        if col in df.columns:
            return col
    return None

def to_numeric_market(series):
    """Преобразует числа и строки с K/M/B и процентами в числа."""
    s = series.astype(str).str.strip()
    s = s.str.replace(",", "", regex=False)
    s = s.str.replace("%", "", regex=False)

    multipliers = {"K": 1e3, "M": 1e6, "B": 1e9, "T": 1e12}
    suffix = s.str.extract(r"([KMBT])\s*$", expand=False).str.upper()
    values = pd.to_numeric(
        s.str.replace(r"[KMBT]\s*$", "", regex=True),
        errors="coerce"
    )

    for letter, multiplier in multipliers.items():
        values.loc[suffix == letter] *= multiplier

    return values

change_col = find_market_col([
    "change_24h", "price_change_percentage_24h",
    "change_24h_pct", "change", "percent_change_24h"
])

volume_col = find_market_col([
    "volume_value", "volume_24h", "total_volume",
    "volume", "daily_volume"
])

name_col = find_market_col([
    "ticker", "symbol", "currency", "name", "coin"
])

observations = []

# Подготовка данных
changes = (
    to_numeric_market(df[change_col]).dropna()
    if change_col else pd.Series(dtype=float)
)

volumes = (
    to_numeric_market(df[volume_col])
    if volume_col else pd.Series(dtype=float)
)

# Используем только положительные значения оборота
valid_volumes = volumes[volumes > 0].dropna()

# 1. НАПРАВЛЕНИЕ РЫНКА
if len(changes) > 0:
    median_change = changes.median()
    mean_change = changes.mean()
    rising_pct = (changes > 0).mean() * 100
    falling_pct = (changes < 0).mean() * 100
    unchanged_pct = (changes == 0).mean() * 100

    if median_change > 0:
        market_direction = "преимущественно растёт"
    elif median_change < 0:
        market_direction = "преимущественно снижается"
    else:
        market_direction = "не имеет выраженного направления"

    observations.append(
        f"1. Направление рынка: рынок {market_direction}; "
        f"медианное изменение за 24 часа — "
        f"{median_change:+.2f}%, растут {rising_pct:.1f}% "
        f"позиций, снижаются {falling_pct:.1f}%. "
        f"Среднее изменение — {mean_change:+.2f}%. "
    )

# 2. КОНЦЕНТРАЦИЯ ЛИКВИДНОСТИ
if len(valid_volumes) > 0:
    total_volume = valid_volumes.sum()
    top3 = valid_volumes.nlargest(min(3, len(valid_volumes)))
    top3_share = top3.sum() / total_volume * 100
    top1_share = top3.iloc[0] / total_volume * 100

    if name_col:
        volume_rank = valid_volumes.sort_values(ascending=False)
        leader_idx = volume_rank.index[0]
        leader_name = str(df.loc[leader_idx, name_col])
    else:
        leader_name = str(valid_volumes.idxmax())

    concentration = (
        "высокая" if top3_share >= 60
        else "умеренная, однако значительная" if top3_share >= 30
        else "низкая"
    )

    observations.append(
        f"2. Концентрация ликвидности: на топ-3 позиции приходится "
        f"{top3_share:.1f}% суточного оборота — концентрация "
        f"{concentration}. Лидер по обороту — {leader_name}; "
        f"его доля составляет {top1_share:.1f}%. "
        "При высокой концентрации динамика нескольких крупных "
        "позиций может определять общую картину оборота."
    )

# 3. СРЕДНЕЕ, МЕДИАНА И ЭКСТРЕМАЛЬНЫЕ ПОЗИЦИИ
if len(changes) > 0:
    median_change = changes.median()
    mean_change = changes.mean()
    mean_median_gap = mean_change - median_change

    if abs(mean_median_gap) < 0.5:
        distribution_note = (
            "среднее близко к медиане, влияние крайних значений "
            "на общий результат относительно невелико"
        )
    elif mean_median_gap > 0:
        distribution_note = (
            "среднее выше медианы: отдельные сильные росты "
            "смещают средний результат вверх"
        )
    else:
        distribution_note = (
            "среднее ниже медианы: отдельные сильные падения "
            "смещают средний результат вниз"
        )

    max_idx = changes.idxmax()
    min_idx = changes.idxmin()

    def position_name(idx):
        if name_col:
            return str(df.loc[idx, name_col])
        return str(idx)

    best_name = position_name(max_idx)
    worst_name = position_name(min_idx)
    best_change = changes.loc[max_idx]
    worst_change = changes.loc[min_idx]

    observations.append(
        f"3. Разброс результатов: среднее изменение — "
        f"{mean_change:+.2f}%, медиана — {median_change:+.2f}%, "
        f"разница — {mean_median_gap:+.2f} п.п.; {distribution_note}. "
        f"Лидер роста — {best_name} ({best_change:+.2f}%), "
        f"лидер падения — {worst_name} ({worst_change:+.2f}%). "
        "Экстремальные движения стоит проверить отдельно, "
        "прежде чем использовать их как ориентир для прогноза"
    )


# 6. Управленческие решения
decisions = []

decisions.append(
""" 1. Перераспределить закупки в пользу категорий с положительной динамикой и устойчивым спросом, постепенно сокращая закупки товаров с отрицательной динамикой. ъ
2. Сохранить приоритет по ликвидным позициям, поддерживать достаточный запас товаров, формирующих значительную долю оборота, но избегать чрезмерной зависимости от нескольких SKU.
3. Провести дополнительную проверку аномалий, отдельно проанализировать товары с максимальным ростом и падением показателей перед принятием решений об увеличении или сокращении закупок.
4. Ввести регулярный мониторинг, еженедельно отслеживать динамику показателей по категориям и SKU, корректируя закупочный план на основе продаж, маржинальности и остатков.
"""
)

# 7. Сборка отчёта
report_lines = [
    "ОТЧЁТ ДЛЯ ЛПР",
    "Тема: мониторинг дефицита товаров на маркетплейсе",
    f"Дата анализа: {REPORT_DATE}",
    f"Источник: витрина «Все товары»; строк в исходном наборе: {n_total}.",
    "",
    "1. ЧТО СОБРАНО",
    f"Получено {n_total} строк данных.",
    "Доступные колонки: " + ", ".join(map(str, report_df.columns)) + ".",
]

if quality_cols:
    report_lines.append(
        "Проверены пропуски в полях: " + ", ".join(quality_cols) + "."
    )

if stock_col:
    report_lines.append(
        f"Найдено поле остатков: {stock_col}."
    )
elif status_col:
    report_lines.append(
        f"Найдено поле статуса наличия: {status_col}."
    )
else:
    report_lines.append(
        "Ограничение: в наборе не обнаружено явного поля остатков "
        "или статуса наличия. Фактический дефицит по этим данным "
        "подтвердить нельзя."
    )

report_lines += [
    "",
    "2. ТРИ КЛЮЧЕВЫХ НАБЛЮДЕНИЯ",
    *observations,
    "",
    "3. УПРАВЛЕНЧЕСКИЕ РЕШЕНИЯ",
    *[
        f"{i}. {decision.split('. ', 1)[-1]}"
        for i, decision in enumerate(decisions, start=1)
    ],
    "",
    "4. ОГРАНИЧЕНИЯ И СЛЕДУЮЩИЙ ШАГ",
    "Отчёт описывает один срез данных, а не динамику во времени. "
    "Для оценки эффекта решений нужно сохранить базовый срез, "
    "повторить сбор после внедрения и сравнить метрики за "
    "сопоставимые периоды. Изменения цены или объёма торгов "
    "не являются прямым доказательством товарного дефицита.",
]

report_text = "\n".join(report_lines)

print(report_text)

````

**Интерпретация.** На первом графике важна не высота столбцов, а концентрация: если на три позиции приходится больше половины оборота, рынок узкий, и ликвидность остальных позиций иллюзорна. На распределении смотрите на медиану и симметрию: длинный хвост влево — отдельные обвалы, а не общий спад. Диаграмма «капитализация — оборот» в логарифмических шкалах выявляет крупные, но неликвидные позиции: именно они создают риск исполнения. ## Ограничения реальной площадки

- **Динамическая подгрузка.** Список товаров может загружаться по мере прокрутки страницы или после выполнения JavaScript. Обычный HTTP-запрос не всегда возвращает полный перечень товаров.
- **Неполнота открытых данных.** Покупательская витрина обычно показывает цену и доступность товара, но не предоставляет полную информацию о складских остатках, поставках и закупочных заказах (аналогично примеру с учебного стенда).
- **Ограничения доступа.** Частота запросов, авторизация, CAPTCHA и условия использования площадки могут ограничивать автоматизированный сбор данных.

Указанные ограничения необходимо учитывать при оценке полноты и достоверности аналитической витрины. Наличие товара на странице также не гарантирует, что он доступен для немедленной отгрузки.

## Возможные легальные источники данных

- **Официальный API маркетплейса для продавцов.** Позволяет получать данные о товарах, доступных остатках, продажах и заказах с периодичностью, предусмотренной API. Подходит для автоматического мониторинга дефицита в собственном ассортименте. Доступность конкретных полей зависит от площадки и прав продавца.
- **API поставщиков.** Позволяет получать текущие остатки у поставщиков, цены и предполагаемые сроки поставки. Эти данные полезны для определения, у кого можно оперативно заказать недостающий товар.
- **Онлайн-фиды поставщиков.** CSV-, XML- или JSON-фиды, предоставляемые через потоковую интеграцию (для мониторинга). При этом частота обновления зависит от поставщика: фид не обязательно отражает изменения в реальном времени.

### Требования к отказоустойчивости

Скрипт обязан переживать без падения ноутбука:

1. **Таймаут ожидания** (`TimeoutError` / `TimeoutException`) — перехват, сохранение скриншота и HTML, понятное сообщение о вероятной причине.
2. **Пустую выборку** — проверка `len(rows) == 0` до перехода к анализу, с указанием, какой селектор не сработал.
3. **Отсутствие оверлея** — отсутствие баннера не должно приводить к исключению.
4. **Частично заполненную строку** — отсутствующая ячейка даёт `None`, а не обрывает цикл.
5. **Нераспознанное значение** — `n/a`, `—`, пустая строка превращаются в `NaN`, но строка остаётся в выборке.
6. **Ограничение цикла догрузки** — максимальное число кликов и условие выхода при отсутствии роста.
